## Project Introduction

The **PAC Regulatory Intelligence Platform** is an AI-assisted decision support application developed to support Regulatory Affairs professionals in evaluating **Post-Approval Changes (PAC)** for pharmaceutical products.

The application combines structured regulatory knowledge, guided decision-making, and Artificial Intelligence to help users identify the appropriate regulatory filing pathway, required supporting documentation, and recommended regulatory actions.

Rather than replacing regulatory professionals, the platform follows a **Human-in-the-Loop** approach, where Artificial Intelligence performs the initial assessment and the Regulatory Affairs expert reviews and validates the recommendation before making the final regulatory decision.

The application currently supports regulatory assessments for the following markets:

* Australia
* Canada
* European Union
* Switzerland

The solution is designed to improve consistency, reduce manual effort, and accelerate regulatory assessments while maintaining regulatory compliance.


# Project Objectives

The primary objectives of the PAC Regulatory Intelligence Platform are:

* Develop an AI-assisted platform for regulatory assessment of post-approval changes.
* Simplify the evaluation of country-specific regulatory requirements.
* Reduce the time required to identify the appropriate regulatory variation pathway.
* Guide users through a structured regulatory decision-making process.
* Improve consistency in regulatory assessments across multiple markets.
* Generate structured regulatory assessment reports for review and documentation.
* Maintain Human-in-the-Loop decision making, ensuring that final regulatory decisions remain under expert supervision.

# Intelligent Regulatory Matching

Once the candidate records have been filtered, the application compares the user's proposed change with regulatory terminology.

The matching engine follows a multi-level strategy.

Level 1
- Exact Match

Level 2
- Normalized Match

Level 3
- Fuzzy Match (RapidFuzz)

This layered approach improves both accuracy and usability while ensuring that regulatory terminology remains the primary decision criterion.

````python
from rapidfuzz import fuzz

query = "Addition manufacturing site"

official = "Addition of manufacturing site"

score = fuzz.ratio(query.lower(), official.lower())

print("Similarity Score:", score)
````

# Matching Decision

The platform evaluates similarity scores using predefined confidence levels.

| Similarity Score | Confidence | Action |
|-----------------|------------|--------|
| 95–100 | Very High | Accept |
| 85–94 | High | Accept |
| 70–84 | Medium | Suggest Alternatives |
| Below 70 | Low | Display Regulatory Guidance |

This confidence-based approach reduces incorrect recommendations while encouraging users to use proper regulatory terminology.

````python
score = 92

if score >= 95:
    decision = "Very High Confidence"

elif score >= 85:
    decision = "High Confidence"

elif score >= 70:
    decision = "Suggest Similar Regulatory Terms"

else:
    decision = "Display Regulatory Guidance"

print(decision)
````

# Guided Regulatory Decision Engine

Not all regulatory changes can be classified based solely on the proposed change description.

Many regulatory pathways depend on additional regulatory conditions.

The PAC Regulatory Intelligence Platform therefore implements a Guided Decision Engine.

This approach reduces user confusion and improves classification accuracy while ensuring regulatory compliance.

# Human-in-the-Loop Decision Making

Artificial Intelligence assists the regulatory professional by identifying the most appropriate regulatory pathway.

However, the final regulatory decision always remains under human expert supervision.

This approach improves:

- Regulatory compliance
- Transparency
- Explainability
- User confidence

The platform is therefore designed to support regulatory experts rather than replace them.
